**categorical variables**

```python
# gender , partner,depedents,Internetservisec,contract, paymentmethod
# encode the data because NN and ML cannot categorical data
# there is no order in my data we do one hot encoding
# in pandas to do one hot encoidng we use the method --> pd.get_dummies
x = pd.get_dummies(x,drop_first=True)
```

```python
x.head()
```

**Train/Test split**

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

**scale the data**

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

**To apply the neural networks convert the data into tensors**

```python
import torch
x_train_tensor = torch.tensor(x_train,dtype=torch.float32)
x_test_tensor = torch.tensor(x_test,dtype=torch.float32)
y_train_tensor = torch.tensor(y_train.values,dtype=torch.float32).reshape(-1,1) # convert the y 1D to 2D format
y_test_tensor = torch.tensor(y_test.values,dtype=torch.float32).reshape(-1,1)
```

```python
print(x_train_tensor.shape)
```

```python
input_features = x_train_tensor.shape[1]
print(input_features)
```

**Build the neural networks**

```python
import torch.nn as nn
class ChurnModel(nn.Module):
  def __init__(self,input_features):
    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(input_features,32),
        nn.ReLU(),
        nn.Linear(32,16),
        nn.ReLU(),
        nn.Linear(16,1),
        nn.Sigmoid()
      )
  def forward(self,x):
    return self.network(x)
```

**create the model**

```python
model = ChurnModel(input_features)
print(model)
```

**claculate the loss**

```python
criterion = nn.BCELoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr = 0.01

)
```

```python
epochs = 100
loss_history = []
for epoch in range(epochs):
  # forward pass
  predictions = model(x_train_tensor)
  #calculate loss
  loss = criterion(predictions,y_train_tensor)
  #gradient --> calculate list loss
  optimizer.zero_grad()
  # Backpropagation
  loss.backward()
  #update weights
  optimizer.step()

  loss_history.append(loss.item())

  if (epoch + 1) % 10 == 0:
    print(
        f"epoch [{epoch+1}/{epochs}]"
        f"loss: {loss.item():.4f}"
    )
```
